### Parcellation

```python
import nibabel as nib
from nilearn import maskers
```

```python
import os
import numpy as np
data_dir = '/Data/HCP/100unrelated/MNINonLinear/'
subs = os.listdir(data_dir)
atlas_path = '/Data/atlas/finn2015/shen_2mm_268_parcellation.nii.gz'
atlas_img = nib.load(atlas_path)
masker = maskers.NiftiLabelsMasker(labels_img=atlas_img, standardize=True, detrend=True, low_pass=0.09, t_r=1)

for sub in subs:
    path_functional_1 = f'/Data/HCP/100unrelated/MNINonLinear/{sub}/rfMRI_REST1_LR/rfMRI_REST1_LR_hp2000_clean.nii.gz'
    path_functional_2 = f'/Data/HCP/100unrelated/MNINonLinear/{sub}/rfMRI_REST2_LR/rfMRI_REST2_LR_hp2000_clean.nii.gz'
    folder_path = f'/Data/HCP/100unrelated/shen268/ts/{sub}/'
    if not(os.path.exists(folder_path)):
        os.makedirs(folder_path)
    if not(os.path.exists(folder_path + f'{sub}_REST1_LR.npy')):
        ts1 = masker.fit_transform(path_functional_1)
        ts2 = masker.fit_transform(path_functional_2)
        np.save(folder_path + f'{sub}_REST1_LR.npy', ts1)
        np.save(folder_path + f'{sub}_REST2_LR.npy', ts2)
```
