# FC from time series

Loads parcellated time series and computes functional connectivity (`fc/whole/whole_corrs.pck`) for:
- HCP 1200 subjects, Schaefer300, `NoGSR_LowPassOnly`
- HCP 100 unrelated: Schaefer300, Glasser379, AAL89, Shen268, Schaefer100

Paths come from `topofinger.config`. Scaffolds are computed separately (`01*_julia`, `scripts/`).

```python
import os
import sys
import sys
sys.path.append("..")
from topofinger import config
from topofinger.hcp_IO import load_ts_schaefer300_dict, save_whole_correlations_dict, load_ts_glasser379_dict, load_ts_aal89_dict, load_ts_shen268_dict, load_scaffold_schaefer300, load_ts_schaefer100_dict, load_ts_custom
from topofinger.utils import apply_threshold_to_nested_dict

import numpy as np
import matplotlib.pyplot as plt

import pickle as pk
from nilearn import plotting




%load_ext autoreload
%autoreload 2
```

# 1200 subjects

```python
data_dir = config.HCP_1200 + "/NoGSR_LowPassOnly/"
ts = "/ts/"
```

```python
from collections import OrderedDict

ts_data = load_ts_custom(data_dir+ts, gsr="NoGSR", bandpass="LowPassOnly")

# ts_data = OrderedDict()
# for sub in ts_data_raw:
#     if len(ts_data_raw[sub]) == 2:
#         ts_data[sub] = ts_data_raw[sub]

print('The dataset is composed of ' + str(len(ts_data.keys())) + ' subjects')

if len(ts_data) > 0:
    first_sub = list(ts_data.keys())[0]
    print('Each subject performed ' + str(len(ts_data[first_sub].keys())) + ' trials')
    print('\n')
    print('Each trial contains ' + str(ts_data[first_sub][0].shape[1]) +' time series (number of brain regions)')
    print('The lenght of each time serie is ' + str(ts_data[first_sub][0].shape[0]) + ' time points')


    # Plot an example of time serie
    fig = plt.figure(figsize=(8,2))
    plt.plot(ts_data[first_sub][0][:,0], 'k-', alpha=0.7)
    plt.box(False)
    plt.xlabel('Time', fontsize=12)
    plt.tight_layout()
```

```python
save_whole_correlations_dict(ts_data, data_dir)
```

```python
with open(data_dir+'fc/whole/whole_corrs.pck', 'rb') as f:
    whole_corrs = pk.load(f)

#sub_number = '100307'    
sub_number = '212318'
session = 0

np.fill_diagonal(whole_corrs[sub_number][session],np.nan)
plotting.plot_matrix(whole_corrs[sub_number][session],cmap='GnBu', title = 'Functional connectivity');
```

# Schaefer300

### Load Data

```python
data_dir = config.HCP_100 + '/schaefer300/no_hp/'
ts = '/ts/LR/'

ts_data = load_ts_schaefer300_dict(data_dir+ts)

print('The dataset is composed of ' + str(len(ts_data.keys())) + ' subjects')
print('Each subject performed ' + str(len(ts_data['103414'].keys())) + ' trials')
print('\n')
print('Each trial contains ' + str(ts_data['103414'][0].shape[1]) +' time series (number of brain regions)')
print('The lenght of each time serie is ' + str(ts_data['103414'][0].shape[0]) + ' time points')
print('\n')


# Plot an example of time serie
fig = plt.figure(figsize=(8,2))
plt.plot(ts_data['103414'][0][:,0], 'k-', alpha=0.7)
plt.box(False)
plt.xlabel('Time', fontsize=12)
plt.tight_layout()
```

### Compute correlation matrices

```python
save_whole_correlations_dict(ts_data, data_dir)
```

```python
with open(data_dir+'fc/whole/whole_corrs.pck', 'rb') as f:
    whole_corrs = pk.load(f)

#sub_number = '212318'    
sub_number = '100307'
session = 0

np.fill_diagonal(whole_corrs[sub_number][session],np.nan)
plotting.plot_matrix(whole_corrs[sub_number][session],cmap='GnBu', title = 'Functional connectivity');
```

```python
fc_th = apply_threshold_to_nested_dict(whole_corrs,1)

plotting.plot_matrix(fc_th[sub_number][session],cmap='GnBu', colorbar = False, title = 'Functional connectivity thresholded');
```

```python
import numpy as np
data_dir = config.HCP_100 + '/schaefer300/no_hp/'
scaffold_dir = 'scaffolds/whole/'
sc = load_scaffold_schaefer300(data_dir+scaffold_dir+'/freq/')

sc_ex = sc[sub_number][session]
sc_ex_bin = np.where(sc_ex > 0, 1, 0)


plotting.plot_matrix(sc_ex_bin,cmap='GnBu', colorbar = False, title = 'Scaffolds');
```

```python
fig, axs = plt.subplots(1, 2, figsize=(12, 6))

# Plot the first matrix
axs[0].imshow(fc_th[sub_number][session], cmap='GnBu')
axs[0].set_title('Functional connectivity thresholded 1%', fontsize=20)

# Plot the second matrix
axs[1].imshow(sc_ex_bin, cmap='GnBu')
axs[1].set_title('Scaffolds', fontsize=20)

# Add colorbars
#cbar1 = fig.colorbar(axs[0].imshow(fc_th[sub_number][session], cmap='GnBu'), ax=axs[0])
#cbar1.set_label('Correlation')
#cbar2 = fig.colorbar(axs[1].imshow(sc_ex_bin, cmap='GnBu'), ax=axs[1])
#cbar2.set_label('Scaffold')

# Adjust spacing between subplots
plt.tight_layout()

# Show the figure
axs[0].set_xticks([])
axs[0].set_yticks([])
axs[1].set_xticks([])
axs[1].set_yticks([])

plt.savefig(
    f"sparsify.png", dpi=250, bbox_inches="tight", facecolor="w"
)
```
